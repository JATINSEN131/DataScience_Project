```python
import pandas as pd
```

```python
raw_df = pd.read_csv('Copy of Online Retail.csv')
```

```python
df = raw_df.copy()
```

```python
print("Original dataset shape:", raw_df.shape)
print("Working copy shape:", df.shape)

print("Datasets identical:", raw_df.equals(df))
```

```python
df['InvoiceDate'] = pd.to_datetime(
    df['InvoiceDate'],
    format='%m/%d/%Y %H:%M'
)
df['InvoiceDate'].min(), df['InvoiceDate'].max()
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df = df.dropna(subset=['CustomerID'])

print("Shape after removing missing CustomerID:", df.shape)
```

```python
df['InvoiceNo'].dtype
```

```python

df = df[~df['InvoiceNo'].str.startswith('C')]
print("Shape after removing cancelled invoices:", df.shape)
```

```python
print("Quantity <= 0:", (df['Quantity'] <= 0).sum())
print("UnitPrice <= 0:", (df['UnitPrice'] <= 0).sum())
```

```python
print("Quantity <= 0 remaining:", (df['Quantity'] <= 0).sum())
print("UnitPrice <= 0 remaining:", (df['UnitPrice'] <= 0).sum())
```

```python

df = df[
    (df['Quantity'] > 0) &
    (df['UnitPrice'] > 0)
].copy()
```

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice']
df[['Quantity', 'UnitPrice', 'TotalAmount']].head()
```

```python
df.to_csv('cleaned_online_retail.csv', index=False)
```
